### Aim
To implement a Markov Decision Process (MDP) using Policy Iteration and Value Iteration algorithms in Python to determine the optimal state values and the optimal policy for maximizing the cumulative reward.

```
import numpy as np
import matplotlib.pyplot as plt
from dp.policy_iteration import run_policy_iteration
from dp.value_iteration_grid import run_value_iteration, NUM_ROWS, NUM_COLS, GOAL_STATE, TRAP_STATE
```

### 1. Policy Iteration Implementation
Policy evaluation iteratively calculates state values until convergence, followed by policy improvement until policy stability.

```
final_values, final_policy = run_policy_iteration()
print("=== Policy Iteration Results ===")
print("State Values :", np.round(final_values, 2))
print("Optimal Policy:", final_policy)
```

### 2. Value Iteration on 2D GridWorld (3x3)
A 2D GridWorld with goal $(0, 2)$, trap $(1, 2)$, step penalty $-1.0$, and wall collision constraints.

```
converged_V, best_policy = run_value_iteration()
print("=== Value Iteration Results ===")
print("Converged Value Grid:")
print(np.round(converged_V, 1))
print("\nOptimal Policy Map:")
for r in range(NUM_ROWS):
    print([best_policy[(r, c)] for c in range(NUM_COLS)])
```

### 3. GridWorld Policy & Value Map Visualization

```
fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(converged_V, cmap='YlGnBu')
plt.colorbar(im, ax=ax, label='State Value V(s)')

for r in range(NUM_ROWS):
    for c in range(NUM_COLS):
        val_text = f"{converged_V[r, c]:.1f}"
        pol_text = best_policy[(r, c)]
        if (r, c) == GOAL_STATE:
            ax.text(c, r, f"GOAL\n+10.0\n({val_text})", ha='center', va='center',
                    color='green', fontweight='bold', fontsize=10, bbox=dict(boxstyle="square", fc="#d4edda", ec="green"))
        elif (r, c) == TRAP_STATE:
            ax.text(c, r, f"TRAP\n-10.0\n({val_text})", ha='center', va='center',
                    color='red', fontweight='bold', fontsize=10, bbox=dict(boxstyle="square", fc="#f8d7da", ec="red"))
        else:
            ax.text(c, r, f"{pol_text}\n{val_text}", ha='center', va='center',
                    color='black', fontweight='bold', fontsize=14)

ax.set_xticks(range(NUM_COLS))
ax.set_yticks(range(NUM_ROWS))
ax.set_title('3x3 GridWorld: Value Heatmap & Policy Arrows', fontweight='bold')
plt.tight_layout()
plt.show()
```

### 4. Conclusion & Findings
Both Policy Iteration and Value Iteration converged to the optimal policy. In the 2D GridWorld, states neighboring the goal exhibit high values and steer toward it, while states adjacent to the trap actively divert away from danger.
